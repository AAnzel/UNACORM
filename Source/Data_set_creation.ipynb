```python
import os
import numpy as np
```

```python
path_root_data = os.path.join('..', 'Data')
path_original_data = os.path.join(path_root_data, 'Original')
path_converted_data = os.path.join(path_root_data, 'Converted')
```

```python
def convert_data_to_binary(path_original_data, path_converted_data):
    for string_file in os.listdir(path_original_data):
        path_original_file = os.path.join(path_original_data, string_file)
        path_converted_file = os.path.join(path_converted_data, string_file)

        # If we have a file, we convert it into bytestream array
        if os.path.isfile(path_original_file):
            with open(path_original_file, 'rb') as fd_file:
                array_bytes = np.fromfile(fd_file, np.dtype('B'))

            np.save(path_converted_file, array_bytes)
    
```

```python
convert_data_to_binary(path_original_data, path_converted_data)
```
